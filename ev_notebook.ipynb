```python
generations = 200

problem = Rastrigin()
optimizer = ParticleSwarmOptimizer(problem=problem,
                                   direction='min',
                                   population_size=100,
                                   dimensions=10,
                                   bounds=[[-5,5]]*10)

fitness = []
for generation in tqdm(range(1,generations+1)):
    best_pos, best_fit = optimizer.evolve()
    fitness.append(best_fit)
    print(f"Generation: {generation} | Best fitness: {best_fit}")
```

```python
best_x, _ = optimizer.get_best()
```

```python
generations = 1000
crossover_operator = BinomialCrossOver(probability=0.9)
selection_operator = GreedySelection()

problem = Rosenbrock()
optimizer = DifferentialEvolution(problem=problem,
                                  direction='min',
                                  population_size=150,
                                  dimensions=10,
                                  bounds=[[-5,5]]*10,
                                  mutation='rand',
                                  crossover_operator=crossover_operator,
                                  selection_operator=selection_operator,
                                  )

fitness = []
for generation in tqdm(range(1,generations+1)):
    best_pos, best_fit = optimizer.evolve()
    fitness.append(best_fit)
    print(f"Generation: {generation} | Best fitness: {best_fit}")
```
