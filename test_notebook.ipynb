```python
k = 2
rng = np.random.default_rng()

def select(population, fitness):
    population_fitness = np.concatenate((population,fitness),axis=1)
    counter = population.shape[0]/k
    mating_pool = []
    while counter != 0:
        tournament_groups = rng.choice(population_fitness, size=(int(population.shape[0]/k), k), replace=False, shuffle=True)
        mating_pool.append(tournament_groups)
        counter-=counter
    mating_pool = np.array(mating_pool).squeeze(axis=0)
    selected_mating_pool = []
    for group in mating_pool:
        best_candidate_idx = 0
        best_candidates = []
        best_candidates.append(group[0])
        for i in range(1,k):
            if group[i][2] > group[best_candidate_idx][2]:
                best_candidates.clear()
                best_candidate_idx = i
                best_candidates.append(group[i])
            elif group[i][2] == group[best_candidate_idx][2]:
                best_candidates.append(group[i][2])
        selected_mating_pool.append(rng.choice(best_candidates, size=1, replace=False, shuffle=True))
    return mating_pool, np.array(selected_mating_pool).squeeze(axis=1)[:,:-1]
```
